```python
# Imports e configuração
import pandas as pd
import matplotlib.pyplot as plt
pd.set_option('display.max_rows', 50)
pd.set_option('display.max_columns', 50)
```

```python
# Leitura dos dados
df = pd.read_csv('F:/projeto-python/olist.csv', low_memory=False)
df.shape
```

```python
# Visão geral
df.head()
```

```python
# Tipos de dados
df.dtypes
```

```python
# Adicionando a coluna de região
df.insert(8, 'region', df['customer_state'].map({
    'SP': 'Sudeste', 'RJ': 'Sudeste', 'MG': 'Sudeste', 'ES': 'Sudeste',
    'RS': 'Sul', 'PR': 'Sul', 'SC': 'Sul',
    'BA': 'Nordeste', 'PE': 'Nordeste', 'CE': 'Nordeste', 'MA': 'Nordeste',
    'PB': 'Nordeste', 'RN': 'Nordeste', 'PI': 'Nordeste', 'AL': 'Nordeste', 'SE': 'Nordeste',
    'DF': 'Centro-Oeste', 'GO': 'Centro-Oeste', 'MT': 'Centro-Oeste', 'MS': 'Centro-Oeste',
    'PA': 'Norte', 'AM': 'Norte', 'TO': 'Norte', 'RO': 'Norte', 'AC': 'Norte', 'AP': 'Norte', 'RR': 'Norte'
}))

df.head()
```

```python
# Ticket medio por região

# Agrupando os pedidos por ID e região e pegando o valor total do pedido. Na base Olist, se um pedido tem mais de 1 item, ele
# cria 1 linha para cada item do pedido
first_value_order = df.groupby(['order_id', 'region'])['payment_value'].first().reset_index()

first_value_order
```

```python
# Usa o first_value_order para agrupar por região para ter o ticket medio

ticket_medio = first_value_order.groupby('region')['payment_value'].mean().reset_index()
ticket_medio.columns = ['Região', 'Ticket Médio']

ticket_medio
```

```python
ticket_medio.sort_values('Ticket Médio', ascending=True).plot(
    x='Região', 
    y='Ticket Médio', 
    kind='barh', 
    title='Ticket Médio por Região'
)
```

```python
# Quantidade de pedidos por estado

# Agrupando os pedidos por ID e região. Na base Olist, se um pedido tem mais de 1 item, ele cria 1 linha para cada item do pedido
pedidos_regiao = df.groupby('region')['order_id'].nunique().reset_index()
pedidos_regiao.columns = ['Região', 'N de Pedidos']
pedidos_regiao.sort_values('N de Pedidos', ascending=False)
```

```python
pedidos_regiao.set_index('Região')['N de Pedidos'].plot(
    kind='pie',
    autopct='%1.1f%%',
    title='Pedidos por Região'
)
plt.ylabel('')
```
